```python
import osmnx as ox
import networkx as nx 

G = ox.graph_from_place("Malabon, Metro Manila, Philippines", network_type="drive")
print(len(G.nodes), "intersections,", len(G.edges), "road segments")
```

```python
ox.plot_graph(G, node_size=0)
```

```python
orig = ox.distance.nearest_nodes(G, X=120.9830, Y=14.6500)
dest = ox.distance.nearest_nodes(G, X=120.9600, Y=14.7500)
```

```python
route = nx.shortest_path(G, orig, dest, weight="length")
ox.plot_graph_route(G, route, node_size=0)
```
